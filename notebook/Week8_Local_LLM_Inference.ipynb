```python
import pandas as pd
import ollama
```

```python
prompts = pd.read_csv("../dataset/prompts.csv")

model = "llama3.2"

responses = []

print("Prompts loaded successfully.")
print(f"Number of prompts: {len(prompts)}")
```

```python
print("Columns:", prompts.columns.tolist())
print(prompts.head())
```

```python
for index, row in prompts.iterrows():

    prompt = row["PROMPT"]

    response = ollama.chat(
        model=model,
        messages=[
            {
                "role": "user",
                "content": prompt
            }
        ]
    )

    answer = response["message"]["content"]

    with open("../dataset/conversation_history.txt", "a", encoding="utf-8") as file:
        file.write(f"PROMPT_ID: {row['PROMPT_ID']}\n")
        file.write(f"DOMAIN: {row['DOMAIN']}\n")
        file.write(f"PROMPT: {prompt}\n")
        file.write(f"RESPONSE: {answer}\n")
        file.write("-" * 80 + "\n")

    responses.append({
        "PROMPT_ID": row["PROMPT_ID"],
        "DOMAIN": row["DOMAIN"],
        "PROMPT": prompt,
        "RESPONSE": answer
    })

print("All responses generated successfully.")
```

```python
result = pd.DataFrame(responses)

result.head()
```

```python
result.to_csv(
    "../dataset/llm_responses.csv",
    index=False
)

print("llm_responses.csv saved successfully.")
```

```python
comparison = result[["PROMPT_ID", "DOMAIN", "PROMPT", "RESPONSE"]]

comparison.to_csv(
    "../dataset/response_comparison.csv",
    index=False
)

print("response_comparison.csv saved successfully.")
```
